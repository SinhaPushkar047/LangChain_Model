```python
from langchain_groq import ChatGroq
from langchain_core.prompts import MessagesPlaceholder,ChatPromptTemplate
from langchain_core.messages import AIMessage,HumanMessage
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model=ChatGroq(
    model='openai/gpt-oss-120b'
)
```

```python
chat_template=ChatPromptTemplate([
    ('system','You are a helpful chatbot name CHATLY for our tech fest OJASS'),
    MessagesPlaceholder(variable_name='chat_history'),
    ('human','{query}')
])
```

```python
chat_hist=[]
```

```python
with open('chat_history.txt',encoding='utf-8') as f:
    chat_hist.extend(f.readlines())
```

```python
chat_hist
```

```python
while True:
    user_input=input('YOU: ')
    print('YOU: ',user_input)
    if user_input.strip() in ['exit','quit']:
        break
    else:
        prompt=chat_template.invoke({'chat_history':chat_hist,'query':user_input})
        response=model.invoke(prompt).content

        print('AI: ',response)

        chat_hist.append(AIMessage(content=response))
        chat_hist.append(HumanMessage(content=user_input))
        with open("chat_history.txt", "a", encoding="utf-8") as f:
            f.write(f"HumanMessage: {user_input}\n")
            f.write(f"AIMessage: {response}\n")
```
